### Assigning the connection string to an environment variable

To securely store your connection string for use in the exercises, you can create an environment variable in DataLab:

1. Go to Environment > Environment variables
2. Click the "..." next to your environment variables > Then click "Edit"
3. Add "MONGODB_URI" as the Name, and assign your connection string `mongodb+srv://...` to the Value.
4. Click Save environment variables
5. Click the "..." again, and select "Connect".

This should restart your session and make this connection string available through `os.environ["MONGODB_URI"]`.

**Enter the connection string for your MongoDB Atlas cluster when prompted.**

```python
import os
from pymongo import MongoClient

MONGODB_URI = os.environ["MONGODB_URI"]
```

**Create a MongoDB Python client and test the cluster connection with a ping.**

```python
# Initialize a MongoDB Python client
mongodb_client = MongoClient(MONGODB_URI, appname="datacamp_text_to_query_course")

# Check the connection to your cluster
ping_result = mongodb_client.admin.command("ping")
print(ping_result)
```
